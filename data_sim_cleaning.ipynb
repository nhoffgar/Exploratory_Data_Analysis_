```python
import pandas as pd
import numpy as np
import os
import sys  
import argparse
from faker import Faker
import random


```

```python
faker = Faker()
products = ["Laptop", "Phone", "Headphones", "Smartwatch", "Tablet", "Gaming Console"]

data = [
    {
        "customer": faker.name(),
        "email": faker.email(),
        "product": random.choice(products),
        "price": round(random.uniform(50, 1500), 2),
        "purchase_date": faker.date_this_year()
    }
    for _ in range(1000)
]

df = pd.DataFrame(data)
df.to_csv("ecommerce_data.csv", index=False)
df.head()
```
